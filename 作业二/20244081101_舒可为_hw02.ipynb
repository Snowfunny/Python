## 字符串的乐趣

回文判断时去掉空格，并把字母统一成小写。字母顺序只看去掉空格后的字母，重复字母是允许的。去掉元音时，`a e i o u` 的大小写都去掉。


```python
def is_palindrome(text):
    cleaned = "".join(ch.lower() for ch in text if ch.isalnum())
    return cleaned == cleaned[::-1]


examples = ["level", "kayak", "pop", "tacocat", "T A C O cat", "rats live on no evil star", "python"]
for text in examples:
    print(repr(text), is_palindrome(text))
```

```python
def is_abecedarian(text):
    letters = [ch.lower() for ch in text if ch != " "]
    for index in range(len(letters) - 1):
        if letters[index] > letters[index + 1]:
            return False
    return True


for text in ["adder", "beet", "dog", "cat", "abcd efgh xyz", "abbey"]:
    print(repr(text), is_abecedarian(text))
```

```python
def remove_vowels(text):
    vowels = set("aeiouAEIOU")
    return "".join(ch for ch in text if ch not in vowels)


for text in ["cat", "audio", "aeiou", "rhythm", "Audio"]:
    print(repr(text), "->", repr(remove_vowels(text)))
```

## 列表的乐趣

`list_reverse` 返回一个新列表，不修改原来的列表。传入的不是列表时抛出 `TypeError`。`is_sorted` 只从左到右看相邻元素一次，相等也算升序。


```python
def list_reverse(items):
    if not isinstance(items, list):
        raise TypeError("参数必须是列表")
    return items[::-1]


numbers = [1, 2, 3]
print(list_reverse(numbers))
print("原列表没有改变:", numbers)
try:
    list_reverse((1, 2, 3))
except TypeError as exc:
    print(type(exc).__name__, exc)
```

```python
def is_sorted(seq):
    for index in range(len(seq) - 1):
        if seq[index] > seq[index + 1]:
            return False
    return True


for seq in ([1, 2, 3], [1, 2, 2, 3], [1, 3, 2], "abc", "cba", []):
    print(repr(seq), is_sorted(seq))
```

## 作为向量的元组

向量用元组保存。整数和浮点数都算有效数值；布尔值虽然是 `int` 的子类，但这里不把它当作向量元素或标量。


```python
def _is_number(value):
    return isinstance(value, (int, float)) and not isinstance(value, bool)


def vec_scalar_mult(t, s):
    if not isinstance(t, tuple) or not _is_number(s) or any(not _is_number(item) for item in t):
        raise TypeError("t 必须是由整数或浮点数组成的元组，s 必须是整数或浮点数")
    return tuple(item * s for item in t)


print(vec_scalar_mult((1, 2, 3), 2))
print(vec_scalar_mult((1.5, 2.0), 0.5))
try:
    vec_scalar_mult([1, 2], 2)
except TypeError as exc:
    print(type(exc).__name__, exc)
```

```python
def vec_inner_product(left, right):
    if not isinstance(left, tuple) or not isinstance(right, tuple):
        raise TypeError("两个参数都必须是元组")
    if any(not _is_number(item) for item in left + right):
        raise TypeError("元组元素必须是整数或浮点数")
    if len(left) != len(right):
        raise ValueError("两个向量的长度必须相同")
    return float(sum(a * b for a, b in zip(left, right)))


print(vec_inner_product((1, 2, 3), (4, 5, 6)))
try:
    vec_inner_product((1, 2), (1, 2, 3))
except ValueError as exc:
    print(type(exc).__name__, exc)
try:
    vec_inner_product([1, 2], (1, 2))
except TypeError as exc:
    print(type(exc).__name__, exc)
```

```python
def check_valid_mx(mx):
    if not isinstance(mx, tuple) or len(mx) == 0:
        return False
    if any(not isinstance(row, tuple) or len(row) == 0 for row in mx):
        return False
    row_length = len(mx[0])
    for row in mx:
        if len(row) != row_length:
            return False
        if any(not _is_number(item) for item in row):
            return False
    return True


candidates = [
    ((1, 2), (3, 4.0)),
    ((1, 2), (3,)),
    ((1, 2), [3, 4]),
    ((1, "a"),),
    (1, 2),
]
for candidate in candidates:
    print(repr(candidate), check_valid_mx(candidate))
```

矩阵是元组套元组：外层每个元素是一行，各行长度相同，元素只能是整数或浮点数。`mx_vec_mult` 计算矩阵乘向量，结果的每一项是对应行与向量的内积。


```python
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise ValueError("矩阵必须是各行等长、元素为整数或浮点数的元组")
    if not isinstance(vec, tuple) or any(not _is_number(item) for item in vec):
        raise TypeError("向量必须是由整数或浮点数组成的元组")
    if len(vec) != len(mx[0]):
        raise ValueError("向量长度必须等于矩阵的列数")
    return tuple(vec_inner_product(row, vec) for row in mx)


print(mx_vec_mult(((1, 2), (3, 4)), (1, 1)))
try:
    mx_vec_mult(((1, 2), (3, 4)), (1,))
except ValueError as exc:
    print(type(exc).__name__, exc)
```

## 更多向量的乐趣

维度和元素都是实例属性。类属性由所有对象共享，一个向量改坐标就会影响其他向量。每个向量有自己的维度和坐标，所以应当写在 `self` 上。

只给出维度时，默认做成同样长度的零向量。维度不是非负整数，或者维度和元素个数对不上，就抛出异常。


```python
class Vector:
    def __init__(self, dimension, elements=None):
        if isinstance(dimension, bool) or not isinstance(dimension, int):
            raise TypeError("维度必须是非负整数")
        if dimension < 0:
            raise ValueError("维度不能为负数")
        if elements is None:
            elements = (0,) * dimension
        elif not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if len(elements) != dimension:
            raise ValueError("维度与元素个数不一致")
        if any(not _is_number(item) for item in elements):
            raise TypeError("元素必须是整数或浮点数")
        self.dimension = dimension
        self.elements = tuple(elements)

    def _require_same_dimension(self, other, action):
        if not isinstance(other, Vector):
            raise TypeError(f"{action}的对象必须是 Vector")
        if self.dimension != other.dimension:
            raise ValueError(f"维度不同，不能{action}")

    def __eq__(self, other):
        self._require_same_dimension(other, "比较")
        return self.elements == other.elements

    def __lt__(self, other):
        self._require_same_dimension(other, "比较")
        return self.elements < other.elements

    def __le__(self, other):
        self._require_same_dimension(other, "比较")
        return self.elements <= other.elements

    def __gt__(self, other):
        self._require_same_dimension(other, "比较")
        return self.elements > other.elements

    def __ge__(self, other):
        self._require_same_dimension(other, "比较")
        return self.elements >= other.elements

    def dot(self, other):
        self._require_same_dimension(other, "计算内积")
        return float(sum(a * b for a, b in zip(self.elements, other.elements)))

    def __mul__(self, other):
        if isinstance(other, Vector):
            self._require_same_dimension(other, "相乘")
            product = tuple(a * b for a, b in zip(self.elements, other.elements))
            return Vector(self.dimension, product)
        if _is_number(other):
            return Vector(self.dimension, tuple(item * other for item in self.elements))
        raise TypeError("只能乘以标量或另一个 Vector")

    def __rmul__(self, other):
        return self.__mul__(other)

    def __repr__(self):
        return f"Vector({self.dimension}, {self.elements})"
```

```python
print(Vector(3))
left = Vector(2, (2, 4))
right = Vector(2, (2, 5))
print("相等:", left == Vector(2, [2, 4]))
print("小于:", left < right)
print("小于等于:", left <= right)
print("大于:", right > left)
print("内积:", left.dot(right))
print("右乘标量:", left * 2)
print("左乘标量:", 2 * left)
print("向量相乘:", Vector(2, (2, 3)) * Vector(2, (4, 5)))

try:
    left < Vector(3, (1, 2, 3))
except ValueError as exc:
    print(type(exc).__name__, exc)
try:
    Vector(-1)
except ValueError as exc:
    print(type(exc).__name__, exc)
try:
    Vector(2, (1,))
except ValueError as exc:
    print(type(exc).__name__, exc)
```

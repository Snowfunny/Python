## 实验1：基本运算、数据类型与字符串

先看四种运算符各自的结果和类型，再检查基本类型、类型转换和字符串操作。


```python
print("8 / 3  =", 8 / 3, type(8 / 3))
print("8 // 3 =", 8 // 3, type(8 // 3))
print("8 % 3  =", 8 % 3, type(8 % 3))
print("2 ** 7 =", 2 ** 7, type(2 ** 7))
```

`/` 是真正的除法，结果是浮点数 `2.666...`。`//` 是整除，只保留商的整数部分，结果是 `2`。`%` 是取余，`8` 除以 `3` 的余数是 `2`。`**` 是幂运算，`2 ** 7` 表示 2 的 7 次方，结果是 `128`。


```python
integer_value = 42
float_value = 3.14
string_value = "python"
bool_value = True

for value in (integer_value, float_value, string_value, bool_value):
    print(repr(value), type(value))
```

```python
print("int('123')   =", int("123"), type(int("123")))
print("float('3.14') =", float("3.14"), type(float("3.14")))
print("int(3.9)      =", int(3.9), type(int(3.9)))

try:
    int("goat")
except ValueError as exc:
    print("int('goat') 失败：", type(exc).__name__, exc)
```

`int("123")` 把数字字符串转成整数 `123`。`float("3.14")` 把小数字符串转成浮点数。`int(3.9)` 不是四舍五入，而是直接丢掉小数部分，所以得到 `3`。`int("goat")` 失败，是因为 `"goat"` 里没有合法的整数字面量，`int()` 无法解释它，于是抛出 `ValueError`。

## 实验2：函数、条件、循环与递归

`score_level` 按左闭右开的分段判断等级，但 100 分属于“优秀”。超出 0〜100 的数值直接报错，避免把非法成绩悄悄算进去。


```python
def score_level(score):
    if score < 0 or score > 100:
        raise ValueError("成绩必须在 0〜100 之间")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"


for score in (100, 90, 89.5, 80, 79, 60, 59, 0):
    print(score, score_level(score))

for score in (-1, 101):
    try:
        score_level(score)
    except ValueError as exc:
        print(score, "->", exc)
```

```python
total_for = 0
for number in range(1, 101):
    total_for += number

total_while = 0
number = 1
while number <= 100:
    total_while += number
    number += 1

print("for  :", total_for)
print("while:", total_while)
print("两种结果相同:", total_for == total_while)
```

```python
def factorial_recursive(n):
    if n < 0:
        raise ValueError("输入必须是非负整数")
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    if n < 0:
        raise ValueError("输入必须是非负整数")
    result = 1
    for factor in range(1, n + 1):
        result *= factor
    return result


for n in (0, 1, 5, 10):
    recursive = factorial_recursive(n)
    iterative = factorial_iterative(n)
    print(f"{n}!  递归={recursive}  迭代={iterative}  相同={recursive == iterative}")
```

递归版在 `n == 0` 时返回 1，这就是约定的 `0! = 1`，同时也是递归的终止条件。迭代版用循环从 1 乘到 n，空乘积同样取 1，所以 `0!` 也是 1。两种写法在 0、1、5、10 上的结果一致。


```python
def summary(values):
    if len(values) == 0:
        raise ValueError("values 不能为空")
    lowest = min(values)
    highest = max(values)
    average = sum(values) / len(values)
    return (lowest, highest, average)


sample_values = [85, 58, 92, 76, 60, 89]
lowest, highest, average = summary(sample_values)
print("最小值:", lowest)
print("最大值:", highest)
print("平均值:", average)
```

```python
def sum_all(*args):
    return sum(args)


print("无参数:", sum_all())
print("单个参数:", sum_all(5))
print("多个参数:", sum_all(1, 2, 3, 4))
```

## 实验4：文件读写、异常处理与模块

数据目录放在笔记本同级的 `exp01_data`。文件是简单的逗号分隔文本，不处理字段内部再出现逗号或引号的情况。无效记录只记录行号和原因，不中断后面的行。


```python
import os

print("工作目录:", os.getcwd())
data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)
students_path = os.path.join(data_dir, "students.txt")

students_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

with open(students_path, "w", encoding="utf-8") as file:
    file.write(students_text)

print("已写入:", students_path)
with open(students_path, encoding="utf-8") as file:
    print(file.read())
```

```python
def parse_student(line):
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量应为 2，实际为 {len(parts)}")

    name = parts[0].strip()
    score_text = parts[1].strip()
    if name == "":
        raise ValueError("姓名为空")

    try:
        score = int(score_text)
    except ValueError as exc:
        raise ValueError(f"成绩无法转换为整数：{score_text}") from exc

    if score < 0 or score > 100:
        raise ValueError(f"成绩超出 0〜100 范围：{score}")

    return {"name": name, "score": score}


valid_records = []
invalid_records = []

with open(students_path, encoding="utf-8") as file:
    for line_number, line in enumerate(file, start=1):
        if line_number == 1:
            continue
        if line.strip() == "":
            continue
        try:
            valid_records.append(parse_student(line))
        except ValueError as exc:
            invalid_records.append((line_number, str(exc)))

print("有效记录:")
for record in valid_records:
    print(record)
print("无效记录:")
for line_number, reason in invalid_records:
    print(f"第 {line_number} 行: {reason}")
```

`w` 模式会清空文件再写入，文件里原来的内容不再保留。`a` 模式从文件末尾接着写，原来的内容还在。下面先用 `w` 写一份会被覆盖的占位文字，再用 `w` 写入真正的统计，最后用 `a` 追加一句说明。


```python
valid_scores = [record["score"] for record in valid_records]
average_score = sum(valid_scores) / len(valid_scores)
pass_rate = sum(score >= 60 for score in valid_scores) / len(valid_scores)

report_path = os.path.join(data_dir, "report.txt")
report_text = (
    f"有效记录数: {len(valid_records)}\n"
    f"无效记录数: {len(invalid_records)}\n"
    f"平均分: {average_score:.2f}\n"
    f"及格率: {pass_rate:.2%}\n"
)

with open(report_path, "w", encoding="utf-8") as file:
    file.write("这段文字会在下一次 w 模式写入时被覆盖。\n")

with open(report_path, "w", encoding="utf-8") as file:
    file.write(report_text)

print("--- 第二次 w 写入之后（占位文字已经消失）---")
with open(report_path, encoding="utf-8") as file:
    print(file.read())

with open(report_path, "a", encoding="utf-8") as file:
    file.write("以上统计由 w 模式写入；本行由 a 模式追加，原有内容保留。\n")

print("--- a 模式追加之后 ---")
with open(report_path, encoding="utf-8") as file:
    print(file.read())
```

## 实验5：类、继承与对象复制

`Student` 保存姓名和成绩。`is_passed()` 以 60 分为及格线。`GraduateStudent` 在此基础上增加研究方向，初始化父类属性时调用 `super().__init__()`，并重写 `__str__()`。


```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"


for record in valid_records:
    student = Student(record["name"], record["score"])
    status = "及格" if student.is_passed() else "不及格"
    print(student, status)
```

```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return (
            f"GraduateStudent(name={self.name}, score={self.score}, "
            f"research_topic={self.research_topic})"
        )


graduate = GraduateStudent("Ivy", 91, "机器学习")
print(graduate)
print("及格" if graduate.is_passed() else "不及格")
print("是 Student 的实例:", isinstance(graduate, Student))
```

直接赋值不会产生新对象，两个变量指向同一份数据。`copy.copy()` 只复制最外层，嵌套的列表仍然共享。`copy.deepcopy()` 会把内部列表也复制一份。每次比较都从新建的原始对象开始，避免上一次修改影响下一次观察。


```python
import copy


def show_copy(title, original, other):
    print(title)
    print("  原对象:", original)
    print("  另一对象:", other)
    print("  是同一个对象:", original is other)
    print("  内部列表是同一个对象:", original["scores"] is other["scores"])


original = {"name": "Alice", "scores": [85, 90]}
assigned = original
assigned["scores"].append(95)
show_copy("直接赋值后修改内部列表", original, assigned)

original = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(original)
shallow["scores"].append(95)
show_copy("浅拷贝后修改内部列表", original, shallow)

original = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(original)
deep["scores"].append(95)
show_copy("深拷贝后修改内部列表", original, deep)
```
